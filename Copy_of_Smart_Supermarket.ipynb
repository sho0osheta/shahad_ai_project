```python

# ==================================================
# 1) إنشاء هيكل البيانات
# الاسم: دلال
# ==================================================

products = {
    "تفاح": {"price": 5.0, "qty": 20},
    "موز": {"price": 7.0, "qty": 50},
    "عصير برتقال المراعي": {"price": 3.5, "qty": 100},
    "جبن كيري": {"price": 12.0, "qty": 50},
    "حليب لتر ونص": {"price": 20.0, "qty": 20},
    "لبن المراعي لتر": {"price": 15.0, "qty": 20},
    "سكر خمسة كيلو": {"price": 40.0, "qty": 20},
    "ملح": {"price": 7.0, "qty": 30},
    "زيت دلال": {"price": 12.0, "qty": 20},
    "ماء نوفا": {"price": 1.5, "qty": 100},
    "بيبسي": {"price": 5.0, "qty": 20},
    "جالكسي": {"price": 5.0, "qty": 60},
    "ليز": {"price": 3.0, "qty": 70},
    "طبق بيض": {"price": 30.0, "qty": 20},
    "نصف طبق بيض": {"price": 15.0, "qty": 10},
    "معجون اسنان كولجيت": {"price": 5.0, "qty": 20},
    "مناديل": {"price": 4.5, "qty": 80}
}

# ==================================================
# 2) عرض قائمة المنتجات
# الاسم: رؤى
# ==================================================

print("-" * 40)
print("📦 قائمة المنتجات المتوفرة:")
print("-" * 40)

for name, info in products.items():
    print(f"🔹 المنتج: {name}")
    print(f"   السعر: {info['price']:.2f} ريال")
    print(f"   الكمية المتاحة: {info['qty']} قطعة")
    print("-" * 25)

# ==================================================
# 3) تجهيز السلة والمتغيرات
# ==================================================

cart = []
subtotal = 0.0

# ==================================================
# 4) إدخال اختيار المستخدم + التحقق من المنتج والكمية
# الأسماء: أميرة + شهد
# ==================================================

while True:
    product_name = input("\nأدخل اسم المنتج الذي ترغب بشرائه (أو اكتب 'انهاء' لطباعة الفاتورة): ").strip()

    if product_name == "انهاء":
        break

    if product_name in products:
        try:
            requested_qty = int(input("أدخل الكمية المطلوبة: "))

            if requested_qty > 0:
                if requested_qty <= products[product_name]["qty"]:

                    # ==================================================
                    # 5) تحديث السلة وخصم الكمية من المخزون
                    # الأسماء: لمى + مشاعل
                    # ==================================================

                    products[product_name]["qty"] -= requested_qty
                    item_price = requested_qty * products[product_name]["price"]
                    subtotal += item_price

                    cart.append({
                        "name": product_name,
                        "qty": requested_qty,
                        "total_price": item_price
                    })

                    print("✅ تمت إضافة المنتج إلى السلة بنجاح.")
                    print(f"💰 سعر هذا المنتج: {item_price:.2f} ريال")
                    print(f"📉 الكمية المتبقية في المخزون: {products[product_name]['qty']}")

                else:
                    print(f"❌ الكمية غير كافية. المتوفر في المخزون هو {products[product_name]['qty']} فقط.")
            else:
                print("❌ يجب أن تكون الكمية أكبر من صفر.")

        except ValueError:
            print("❌ خطأ: الرجاء إدخال رقم صحيح للكمية.")
    else:
        print("❌ عذرًا، هذا المنتج غير موجود في النظام.")

# ==================================================
# 6) حساب المجموع والضريبة والإجمالي النهائي
# الأسماء: رنا + سحر
# ==================================================

if len(cart) > 0:
    vat = subtotal * 0.15
    final_total = subtotal + vat

    # ==================================================
    # 7) الفاتورة النهائية
    #+ كلثوم - الاسم: وجدان
    # ==================================================

    print("\n" + "=" * 40)
    print("🧾 الفاتورة النهائية 🧾")
    print("=" * 40)

    for item in cart:
        print(f"- {item['name']} (الكمية: {item['qty']}) = {item['total_price']:.2f} ريال")

    print("-" * 40)
    print(f"المجموع الفرعي: {subtotal:.2f} ريال")
    print(f"ضريبة القيمة المضافة (15%): {vat:.2f} ريال")
    print("=" * 40)
    print(f"الإجمالي النهائي المستحق: {final_total:.2f} ريال")
    print("=" * 40)
    print("شكرًا لتسوقكم معنا!")

else:
    print("\nلم يتم شراء أي منتجات. نراكم قريبًا!")
```
